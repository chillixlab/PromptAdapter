```python
from prompt_adapter.evaluator.run import Evaluator
from datasets import Dataset

evaluator = Evaluator(evaluate_llm_name="gpt-5.2-chat")

data = {
    "question": ["What is the capital of France?", "電通総研の大沢とはどんな人？"],
    "answer": ["Paris", "大沢は電通総研のAIの営業です。"],
    "contexts": [
        ["France is in Europe. Paris is its capital."],
        ["大沢は電通総研のAIのリサーチャーです。"],
    ],
    "ground_truth": ["Paris", "大沢は電通総研のAIのリサーチエンジニアです。"],
}
# データセットの作成
dataset = Dataset.from_dict(data)

# RAGAS評価の実行
result = evaluator.evaluate(model_name="gpt-4o", dataset=dataset)
```
